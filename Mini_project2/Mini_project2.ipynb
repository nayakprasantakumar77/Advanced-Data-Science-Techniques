```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler
```

```python
data={
    'Age':[25,45,31,52,92]
}
df=pd.DataFrame(data)
print(df)
```

```python
scaler= MinMaxScaler()
```

```python
scaled=scaler.fit_transform(df[['Age']])
```

```python
df_scaled=pd.DataFrame(scaled, columns=['Age_Scaled'])
```

```python
print(df_scaled)
```
